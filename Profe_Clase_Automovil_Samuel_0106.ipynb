```python
# Clase Automovil, atributos, marca, placas, precio
# Funciones leerdatos(), mostardatos()
# crear 3 objetos carro1, carro2, carro3

class Automovil:
  def __init__(self, marca, placas, precio):
      self.marca = marca
      self.placas = placas
      self.precio = precio

  def leerdatos(self):
      self.marca = input("Ingrese la marca del Automovil: ")
      self.placas = input("Ingrese las placas del Automovil: ")
      self.precio = input("Ingrese el precio del Automovil: ")
      return self.marca, self.placas, self.precio

  def mostardatos(self):
      print("Marca: ", self.marca)
      print("Placas: ", self.placas)
      print("Precio: ", self.precio)
      return self.marca, self.placas, self.precio
# fin de la clase automovil

print("carro numero 1")
carro1 = Automovil("","","") # creando el objeto carro1
carro1.leerdatos()
carro1.mostardatos()
print("_*_*_*_*_*_*_*_*_*_*_*_*_*_")
print("carro numero 2")
carro2 = Automovil("","","") # creando el objeto carro3
carro2.leerdatos()
carro2.mostardatos()
print("_*_*_*_*_*_*_*_*_*_*_*_*_*_")
print("carro numero 3")
carro3 = Automovil("","","") # creando el objeto carro3
carro3.leerdatos()
carro3.mostardatos()
print("programa elaborado por Samuel Nicola NL = 40")

```
